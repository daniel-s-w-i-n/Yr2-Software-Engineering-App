```python
# Title and Introduction

"""
# Initial Data Check for PV Data Analysis
This notebook performs an initial check on the harvested PV data to ensure its integrity and to understand its structure.
"""
```

```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-darkgrid')
```

```python
# Load the Data

data = pd.read_excel('../data/Weather Data 2014-11-30.xlsx', sheet_name='Sheet1')
```

```python
# Preliminary Data Inspection

print(data.head())

print(data.info())
```

```python
# Check for Missing Values

missing_values = data.isnull().sum()
print(missing_values)
```

```python
# Basic Statistical Overview

print(data.describe())
```

```python
# Initial Data Quality Assessment
# Duplicate entries check

duplicates = data.duplicated().sum()
print(f"Number of duplicate entries: {duplicates}")

# Anomalies or outliers check

for column in data.select_dtypes(include=np.number).columns:
    plt.figure(figsize=(10, 4))
    sns.boxplot(data[column])
    plt.title(f'Boxplot of {column}')
    plt.show()
```

```python
# Categorical Data Overview

for column in data.select_dtypes(include='object').columns:
    print(f"Unique values for {column}:")
    print(data[column].value_counts())
    print("\n")
```

```python
# Data Distribution Visualization

for column in data.select_dtypes(include=np.number).columns:
    plt.figure(figsize=(10, 4))
    sns.histplot(data[column], kde=True, bins=30)
    plt.title(f'Distribution of {column}')
    plt.show()
```

```python
# Correlation Analysis (if applicable)

correlation_matrix = data.corr()
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```
